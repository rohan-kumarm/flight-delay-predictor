```python
import pandas as pd

df = pd.read_csv("flights_jan2024.csv")   # replace with the actual filename
print(df.shape)
print(df.columns.tolist())
print(df.head())
```

```python
keep_cols = [
    "FlightDate", "Month", "DayOfWeek",
    "Reporting_Airline",
    "OriginAirportID", "OriginAirportSeqID", "OriginCityMarketID", "Origin",
    "DestAirportID", "DestAirportSeqID", "DestCityMarketID", "Dest",
    "CRSDepTime", "CRSArrTime",
    "ArrDel15", "ArrDelay",
    "Cancelled", "Diverted",
    "Distance", "CRSElapsedTime",
    "CarrierDelay", "WeatherDelay", "NASDelay", "SecurityDelay", "LateAircraftDelay"
]

df_clean = df[keep_cols]
print(df_clean.shape)
df_clean.head()
```

```python
airlines_df = df_clean[["Reporting_Airline"]].drop_duplicates()
airlines_df = airlines_df.rename(columns={"Reporting_Airline": "airline_code"})
print(airlines_df.shape)
airlines_df.head()
```

```python
origin_airports = df_clean[["OriginAirportID", "Origin", "OriginCityMarketID"]].rename(
    columns={"OriginAirportID": "airport_id", "Origin": "airport_code", "OriginCityMarketID": "city_market_id"}
)

dest_airports = df_clean[["DestAirportID", "Dest", "DestCityMarketID"]].rename(
    columns={"DestAirportID": "airport_id", "Dest": "airport_code", "DestCityMarketID": "city_market_id"}
)

airports_df = pd.concat([origin_airports, dest_airports]).drop_duplicates(subset="airport_id")
print(airports_df.shape)
airports_df.head()
```

```python
from sqlalchemy import create_engine

engine = create_engine("postgresql+psycopg2://localhost/flight_delay_db")

airlines_df.to_sql("airlines", engine, if_exists="append", index=False)
airports_df.to_sql("airports", engine, if_exists="append", index=False)
df_clean.to_sql("flights_staging", engine, if_exists="replace", index=False)

print("Done loading")
```
